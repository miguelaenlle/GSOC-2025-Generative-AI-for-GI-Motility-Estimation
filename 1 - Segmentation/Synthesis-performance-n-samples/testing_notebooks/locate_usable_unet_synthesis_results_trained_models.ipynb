```python
import os
import pandas as pd
```

```python
high_quality_cases = pd.read_csv('high_quality_cases.csv', index_col = 0)
```

```python
trained_model_path = '/home/miguel/GI/1 - Segmentation/UNet-and-Synthesis-Results/TrainedModels'
trained_model_folders = os.listdir(trained_model_path)
```

```python
case_ids = set(high_quality_cases['case_id'].unique())
```

```python
existing_trained_models = set(os.listdir('TrainedModels'))
```

```python
high_quality_cases[high_quality_cases['case_id'] == 'case148_day20']
```

```python
high_quality_cases
```

```python
matching_folders = []   
for trained_model_folder in trained_model_folders:
    case_day = trained_model_folder.split('_slice')[0]
    if case_day in case_ids:
        if os.path.exists(os.path.join(trained_model_path, trained_model_folder, 'scale_factor=0.750000,alpha=10', 'Gs.pth')):
            if trained_model_folder not in existing_trained_models:
                matching_folders.append(trained_model_folder)
```

```python
# Transfer to the trained models folder
import shutil
for folder in matching_folders:
    src = os.path.join(trained_model_path, folder)
    dst = os.path.join('TrainedModels', folder)
    shutil.copytree(src, dst)
```
