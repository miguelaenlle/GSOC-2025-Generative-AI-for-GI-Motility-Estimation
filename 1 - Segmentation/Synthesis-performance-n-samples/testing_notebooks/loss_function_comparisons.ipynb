```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
losses_data = [
    {
        'loo_validation_subject': 'FD-027',
        'file': 'losses-FD-027-2025-08-25_01-49-01.csv'
    },
    {
        'loo_validation_subject': 'FD-029',
        'file': 'losses-FD-029-2025-08-25_18-04-59.csv'
    },
    {
        'loo_validation_subject': 'FD-030',
        'file': 'losses-FD-030-2025-08-26_09-02-06.csv'
    },
    {
        'loo_validation_subject': 'FD-031',
        'file': 'losses-FD-031-2025-08-25_18-54-24.csv'
    },
    {
        'loo_validation_subject': 'FD-032',
        'file': 'losses-FD-032-2025-08-25_19-43-46.csv'
    }
]
```

```python
TMP_OUTPUT_FOLDER = '/home/miguel/GI/1 - Segmentation/Synthesis-performance-n-samples/tmp_output'
```

```python
for loss_data in losses_data:
    df = pd.read_csv(f"{TMP_OUTPUT_FOLDER}/{loss_data['file']}")
    plt.title(f"Loss Curves  - LOO Subject = {loss_data['loo_validation_subject']}")
    plt.plot(df['step'], df['val_loss'], label=f"Val Loss - {loss_data['loo_validation_subject']}")
    plt.plot(df['step'], df['train_loss'], label=f"Train Loss - {loss_data['loo_validation_subject']}")
    plt.xlabel("Training Steps")
    plt.ylabel("Loss")
    plt.legend()
    plt.show()
```
