```python
import os
import random
import cv2
import matplotlib.pyplot as plt

def plot_random_overlays(folder_path, n=5):
    """
    Randomly select n images ending with '_image.png' and overlay their
    corresponding '_mask.png' and '-predicted.png' on top.
    """
    # 1) Gather all your “image” files
    all_files = os.listdir(folder_path)
    image_files = [f for f in all_files if f.endswith('_image.png') and '030' in f]

    # 2) Sample n of them
    selected = random.sample(image_files, min(n, len(image_files)))

    for image_fname in selected:
        mask_fname = image_fname.replace('_image.png', '_mask.png')
        pred_fname = image_fname.replace('-original_image.png', '-predicted.png')

        # full paths
        img_path  = os.path.join(folder_path, image_fname)
        msk_path  = os.path.join(folder_path, mask_fname)
        prd_path  = os.path.join(folder_path, pred_fname)

        # 3) Load (as grayscale for mask/pred)
        img = cv2.imread(img_path)[..., ::-1]                      # BGR→RGB
        msk = cv2.imread(msk_path,  cv2.IMREAD_GRAYSCALE)
        prd = cv2.imread(prd_path, cv2.IMREAD_GRAYSCALE)
        # return prd, prd_path
        # 5) Plot with overlays
        fig, ax = plt.subplots(figsize=(6,6))

        ax.imshow(img, cmap='gray')
        ax.imshow(msk, cmap='Reds',   alpha=0.4, vmin=0, vmax=1)
        ax.imshow(prd, cmap='Blues',  alpha=0.4, vmin=0, vmax=1)
        ax.set_title(image_fname)
        ax.axis('off')
        plt.show()
```

```python
prd, prd_path = plot_random_overlays(pred_folder)
```
