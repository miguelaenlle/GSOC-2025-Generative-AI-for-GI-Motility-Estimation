Prints all images, individual organ masks, and all masks overlayed

```python
# overlay masks on the image and display the result
def display_image_with_mask(image_path, mask_path):
    # get the paths so we can pic some images from the folder
    print(f"Image path: {image_path}")
    print(f"Mask path: {mask_path}")

    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE) # load img
    image = cv2.normalize(image, None, 0, 255, cv2.NORM_MINMAX) # normalize for better viz
    mask = cv2.imread(mask_path) # load mask
    if mask is None: # is mask there?
        return

    # split mask into individual channels
    large_bowel_mask = mask[:,:,0]
    small_bowel_mask = mask[:,:,1]
    stomach_mask = mask[:,:,2]

    # do all three masks contain non-zero vals?
    if not (np.any(large_bowel_mask) and np.any(small_bowel_mask) and np.any(stomach_mask)):
        return

    # make colored masks
    large_bowel_color = np.zeros_like(mask)
    small_bowel_color = np.zeros_like(mask)
    stomach_color = np.zeros_like(mask)

    large_bowel_color[:,:,1] = large_bowel_mask  # green channel = large bowel
    small_bowel_color[:,:,2] = small_bowel_mask  # blue channel = small bowel
    stomach_color[:,:,0] = stomach_mask  # red channel = stomach

    # overlay masks on original imgs
    overlay_large_bowel = cv2.addWeighted(cv2.merge([image, image, image]), 0.7, large_bowel_color, 0.3, 0)
    overlay_small_bowel = cv2.addWeighted(cv2.merge([image, image, image]), 0.7, small_bowel_color, 0.3, 0)
    overlay_stomach = cv2.addWeighted(cv2.merge([image, image, image]), 0.7, stomach_color, 0.3, 0)

    # combine all masks into one overlay for best viz
    combined_masks = cv2.addWeighted(large_bowel_color, 0.3, small_bowel_color, 0.3, 0)
    combined_masks = cv2.addWeighted(combined_masks, 1, stomach_color, 0.3, 0)
    overlay_all = cv2.addWeighted(cv2.merge([image, image, image]), 0.7, combined_masks, 0.3, 0)

    plt.figure(figsize=(20, 10))

    plt.subplot(1, 5, 1)
    plt.title('Image')
    plt.imshow(image, cmap='gray')
    plt.axis('off')

    plt.subplot(1, 5, 2)
    plt.title('Large Bowel Mask Overlay')
    plt.imshow(overlay_large_bowel)
    plt.axis('off')

    plt.subplot(1, 5, 3)
    plt.title('Small Bowel Mask Overlay')
    plt.imshow(overlay_small_bowel)
    plt.axis('off')

    plt.subplot(1, 5, 4)
    plt.title('Stomach Mask Overlay')
    plt.imshow(overlay_stomach)
    plt.axis('off')

    plt.subplot(1, 5, 5)
    plt.title('All Masks Overlay')
    plt.imshow(overlay_all)
    plt.axis('off')

    plt.show()

for index, row in df.iterrows():
    # img path from kaggle
    image_relative_path = os.path.relpath(row['image_path'], '/kaggle/input/uw-madison-gi-tract-image-segmentation/train')
    image_path = os.path.join(images_base_dir, image_relative_path)

    # mask path from kaggle
    mask_relative_path = os.path.relpath(row['mask_path'], '/kaggle/input/uwmgi-mask-dataset/png/uw-madison-gi-tract-image-segmentation/train')
    mask_path = os.path.join(masks_base_dir, mask_relative_path)

    # time to view!
    if os.path.exists(image_path) and os.path.exists(mask_path):
        display_image_with_mask(image_path, mask_path)
```

```python
# calc the brightness of an image
def calculate_brightness(image_path):
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    return np.mean(image)

def display_images_with_brightness(image_paths, brightness_values):
    plt.figure(figsize=(20, 10))
    for i, (image_path, brightness) in enumerate(zip(image_paths, brightness_values)):
        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        plt.subplot(1, len(image_paths), i+1)
        plt.title(f'Brightness: {brightness:.2f}')
        plt.imshow(image, cmap='gray')
        plt.axis('off')
    plt.show()

# calc brightness for each image
df['brightness'] = df['image_path'].apply(lambda x: calculate_brightness(os.path.join(images_base_dir, os.path.relpath(x, '/kaggle/input/uw-madison-gi-tract-image-segmentation/train'))))

# filter DF for just images with brightness between 1.5 and 3.0
filtered_df = df[(df['brightness'] >= 1.5) & (df['brightness'] <= 3.0)]

# time to view!
sample_df = filtered_df.sample(5)  # how many do you want to view?
sample_image_paths = sample_df['image_path'].apply(lambda x: os.path.join(images_base_dir, os.path.relpath(x, '/kaggle/input/uw-madison-gi-tract-image-segmentation/train'))).tolist()
sample_brightness_values = sample_df['brightness'].tolist()
display_images_with_brightness(sample_image_paths, sample_brightness_values)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm
from PIL import Image

def calculate_brightness_and_contrast(image_path):
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    brightness = np.mean(image)
    contrast = np.std(image)
    return brightness, contrast

def display_images_with_metrics(image_paths, metrics_values):
    plt.figure(figsize=(20, 10))
    for i, (image_path, metrics) in enumerate(zip(image_paths, metrics_values)):
        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        plt.subplot(1, len(image_paths), i+1)
        plt.title(f'Brightness: {metrics[0]:.2f}\nContrast: {metrics[1]:.2f}')
        plt.imshow(image, cmap='gray')
        plt.axis('off')
    plt.show()

def normalize_image(image):
    # Normalize image to the range 0-255
    normalized_image = cv2.normalize(image, None, 0, 255, cv2.NORM_MINMAX)
    return normalized_image

def rle2mask(mask_rle, shape):
    if mask_rle == '':
        return np.zeros(shape[0] * shape[1], dtype=np.uint8).reshape(shape)
    s = mask_rle.split(" ")
    starts, lengths = [np.asarray(x, dtype=int) for x in (s[0:][::2], s[1:][::2])]
    ends = starts + lengths - 1
    img = np.zeros(shape[0] * shape[1], dtype=np.uint8)
    for low, high in zip(starts-1, ends):
        img[low : high] = 1
    return img.reshape(shape)

def pathToInfo(path):
    try:
        path_segments = path.split('/')
        if len(path_segments) < 8:
            raise ValueError("Path does not contain enough segments")
        
        caseDay = path_segments[5]
        slice_no = path_segments[7].split('_')[1]
        id = caseDay + '_slice_' + slice_no
        
        details = path_segments[7].split('_')
        if len(details) < 6:
            raise ValueError("Path details segment does not contain enough parts")
        
        height = details[2]
        width = details[3]
        h_spacing = details[4]
        w_spacing = details[5][:-4]  # Remove the file extension
        
        return pd.Series([id, height, width, h_spacing, w_spacing])
    
    except IndexError:
        print(f"IndexError processing path: {path}")
        return pd.Series([None, None, None, None, None])
    except ValueError as ve:
        print(f"ValueError: {ve} for path: {path}")
        return pd.Series([None, None, None, None, None])

# Load train.csv
train_csv_path = '/Users/elizabethnemeti/Desktop/uw-madison-gi-tract-image-segmentation/train.csv'
df = pd.read_csv(train_csv_path)

# Get train images paths
train_images = glob('/Users/elizabethnemeti/Desktop/uw-madison-gi-tract-image-segmentation/train/*/*/*/*')

# Create a DataFrame from the image paths
path_details_df = pd.DataFrame(train_images, columns=['path'])

# Apply the function to extract details and create columns for id, height, width, h_spacing, w_spacing
path_details_df[['id', 'height', 'width', 'h_spacing', 'w_spacing']] = path_details_df['path'].apply(lambda x: pathToInfo(x))

# Merge the new DataFrame with the existing one
df = df.merge(path_details_df, on='id')

# Define folders for saving filtered images and labels
output_folder = 'filtered_images_and_labels'
os.makedirs(output_folder, exist_ok=True)

# Define the thresholds
brightness_threshold = 2.5
contrast_threshold = 0.5

# Add brightness and contrast to DataFrame
df[['brightness', 'contrast']] = df['path'].apply(lambda x: pd.Series(calculate_brightness_and_contrast(x)))

# Filter the dataset to only include images above the brightness and contrast threshold
filtered_df = df[(df['brightness'] >= brightness_threshold) & (df['contrast'] >= contrast_threshold)]

# Display a sample of 5 images being saved
sample_df = filtered_df.sample(5)
sample_image_paths = sample_df['path'].tolist()
sample_metrics_values = sample_df[['brightness', 'contrast']].values.tolist()
display_images_with_metrics(sample_image_paths, sample_metrics_values)

# Save filtered images and labels
for index, row in tqdm(filtered_df.iterrows(), total=filtered_df.shape[0], desc="Saving images and labels"):
    image_path = row['path']
    if os.path.exists(image_path):
        # Normalize and save the image
        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        normalized_image = normalize_image(image)
        base_filename = os.path.basename(image_path).replace('.png', '')
        
        # Create sub-folder for each image
        sub_folder = os.path.join(output_folder, base_filename)
        os.makedirs(sub_folder, exist_ok=True)
        
        new_image_path = os.path.join(sub_folder, f'{base_filename}_image.png')
        cv2.imwrite(new_image_path, normalized_image)
        
        # Decode and save the corresponding label
        mask_rle = row['segmentation']
        shape = (int(row['height']), int(row['width']))
        mask = rle2mask(mask_rle, shape)
        new_mask_path = os.path.join(sub_folder, f'{base_filename}_mask.png')
        Image.fromarray(mask * 255).save(new_mask_path)
```
