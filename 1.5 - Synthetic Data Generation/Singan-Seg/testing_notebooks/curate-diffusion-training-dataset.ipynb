```python
import os
import shutil
```

```python
cv_subjects = ['FD-027', 'FD-029', 'FD-030', 'FD-031', 'FD-032']
```

```python
random_samples_folder_path = '/home/miguel/GI/1.5 - Synthetic Data Generation/Singan-Seg/Output/RandomSamples'
full_dataset_singan_augmented_folder_path = '/home/miguel/GI/0 - Data Exploration & Analysis/GI-Roberta/gi-roberta-dataset/full_dataset_augmented'
```

```python
for folder in os.listdir(random_samples_folder_path):
    curr_cv_subject = None
    for cv_subject in cv_subjects:
        if cv_subject in folder:
            curr_cv_subject = cv_subject
            break
    if not curr_cv_subject:
        continue
    
    folder_path = os.path.join(random_samples_folder_path, folder, 'gen_start_scale=0')
    for file in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file)
        base_file = os.path.basename(file_path)
        new_synthetic_image_filename = folder + "_" + base_file
        shutil.copy(
            file_path,
            os.path.join(full_dataset_singan_augmented_folder_path, new_synthetic_image_filename)
        )
```
