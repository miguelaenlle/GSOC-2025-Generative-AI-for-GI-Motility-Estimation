```python
import os
import random
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
finetuning_random_samples_no_train_mask = '/home/miguel/GI/1.5 - Synthetic Data Generation/Singan-Seg/Output/FinetuningSamples/RandomSamples'
random_samples_with_train_mask = '/home/miguel/GI/1.5 - Synthetic Data Generation/Singan-Seg/Output/RandomSamples'
```

```python
for folder in os.listdir(finetuning_random_samples_no_train_mask):
    random_folder = os.path.join(random_samples_with_train_mask, folder, 'gen_start_scale=0')
    finetuning_folder = os.path.join(finetuning_random_samples_no_train_mask, folder, 'gen_start_scale=0')

    if os.path.exists(random_folder) and os.path.exists(finetuning_folder):
        # list all PNGs
        random_files   = [f for f in os.listdir(random_folder)   if f.lower().endswith('.png') and 'img' in f]
        finetune_files = [f for f in os.listdir(finetuning_folder) if f.lower().endswith('.png') and 'img' in f]

        # sample up to 5 of each
        random_samples   = random.sample(random_files,   min(5, len(random_files)))
        finetune_samples = random.sample(finetune_files, min(5, len(finetune_files)))

        # set up 2 rows × 5 cols
        fig, axes = plt.subplots(2, 5, figsize=(15, 6))

        # top row: random (with mask)
        for i, fname in enumerate(random_samples):
            img = mpimg.imread(os.path.join(random_folder, fname))
            axes[0, i].imshow(img)
            axes[0, i].axis('off')
            axes[0, i].set_title(f'With Mask\n{fname}', fontsize=8)

        # bottom row: finetune (no mask)
        for i, fname in enumerate(finetune_samples):
            img = mpimg.imread(os.path.join(finetuning_folder, fname))
            axes[1, i].imshow(img)
            axes[1, i].axis('off')
            axes[1, i].set_title(f'No Mask\n{fname}', fontsize=8)

        plt.tight_layout()
        plt.show()
```
