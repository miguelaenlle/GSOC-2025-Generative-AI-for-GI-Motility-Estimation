```python
finetuning_dataset_path = '/home/miguel/GI/1.5 - Synthetic Data Generation/Singan-Seg/unet_singan_augmented_datasets_with_finetuning_num_images_10'
```

```python
import os
import cv2
import shutil
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
finetuning_data_RGBA = '/home/miguel/GI/1.5 - Synthetic Data Generation/Singan-Seg/Input/finetuning-data-RGBA'
```

```python
shutil.rmtree(finetuning_data_RGBA, ignore_errors=True)
os.makedirs(finetuning_data_RGBA, exist_ok=True)
```

```python
for dataset_path in os.listdir(finetuning_dataset_path):
    dataset_full_path = os.path.join(finetuning_dataset_path, dataset_path)
    for cv_subject in os.listdir(dataset_full_path):
        cv_subject_finetuning_ds_full_path = os.path.join(dataset_full_path, cv_subject, 'finetuning')
        for sample in os.listdir(cv_subject_finetuning_ds_full_path):
            if sample.endswith('.png') and 'image' in sample:
                sample_full_path = os.path.join(cv_subject_finetuning_ds_full_path, sample)
                # Transform into a 4D image
                image = Image.open(sample_full_path)
                image = image.convert("RGBA")
                image = np.array(image)
                # Save to finetuning data RGBA folder
                new_rgba_image_path = os.path.join(finetuning_data_RGBA, sample)
                cv2.imwrite(new_rgba_image_path, image)
```
